# **🧩 Organización del código en Módulos**

Hasta ahora, hemos escrito todo nuestro código en un único archivo o celda. Esto funciona bien para scripts pequeños, pero imagina programar un videojuego o una herramienta de hacking completa en un solo archivo de 5.000 líneas. Sería imposible de leer y corregir.

Para solucionar esto, Python usa **Módulos**.

## **1️⃣ ¿Qué es un módulo?**

En Python, la definición es sorprendentemente simple:

> **Cualquier archivo con extensión `.py` es un módulo.**

Si tienes un archivo llamado `hack_tools.py`, ese es el módulo `hack_tools`. Dentro de ese archivo puedes guardar:

* **Funciones** (ej: `escanear_ip()`)
* **Clases** (ej: `class Troyano:`)
* **Variables** (ej: `puerto_defecto = 80`)

### **🛠️ Analogía: La Caja de Herramientas**

Imagina tu código como un taller mecánico.

* **Sin módulos:** Tienes todas las herramientas (martillo, sierra, destornillador, llaves) tiradas por el suelo en un montón gigante. Encontrar algo es un caos.
* **Con módulos:** Tienes cajas etiquetadas. Una caja "Electricidad", una caja "Motor", una caja "Ruedas". Sabes exactamente dónde buscar.

## **2️⃣ Ventajas de usar Módulos**

Según la documentación oficial y las buenas prácticas, dividir tu código te da tres superpoderes:

1. **Mantenibilidad:** Si falla la conexión a la base de datos, vas al archivo `database.py`. No tienes que buscar entre miles de líneas de código que no tienen nada que ver.
2. **Reutilización:** Si escribes una función genial para calcular contraseñas en un módulo, puedes usar ese mismo archivo en 10 proyectos diferentes sin tener que copiar y pegar el código.
3. **Espacios de Nombres (Namespaces):** (Este es importante 👇)

### **¿Qué es el Espacio de Nombres?**

Es lo que evita que las variables se "peleen" entre sí.
Imagina que tienes dos módulos: `red.py` y `archivo.py`.

* En `red.py` tienes una función `abrir()` (para abrir una conexión).
* En `archivo.py` tienes una función `abrir()` (para abrir un fichero).

¿Cómo sabe Python cuál usar?
Gracias a los módulos, tienen "apellidos" distintos:

* `red.abrir()`
* `archivo.abrir()`

Así nunca hay conflictos, aunque uses los mismos nombres.

## **3️⃣ Creando nuestro primer módulo**

Vamos a ver cómo se ve esto en la práctica. Crearemos un módulo muy sencillo llamado `operaciones.py`.

> **Nota:** Usaremos un comando especial de Jupyter (`%%writefile`) para crear el archivo desde aquí. Si estuvieras en VSCode, simplemente crearías un archivo nuevo y escribirías el código dentro.

```python
%%writefile operaciones.py

# Este es el contenido de nuestro módulo "operaciones"
# Aquí definimos funciones que queremos reutilizar

mensaje_bienvenida = "Hola desde el módulo de operaciones"

def sumar(a, b):
    return a + b

def restar(a, b):
    return a - b
```

**Salida esperada:**
`Writing operaciones.py`

¡Listo! Acabamos de crear un archivo real en tu disco duro. Ahora Python ya reconoce `operaciones` como un módulo válido.

## **4️⃣ Probando que funciona**

Aunque profundizaremos en la importación en la próxima clase, vamos a ver una prueba rápida para confirmar que podemos acceder a lo que acabamos de crear.

```python
import operaciones

# Accedemos a la variable del módulo
print(operaciones.mensaje_bienvenida)

# Usamos la función del módulo
resultado = operaciones.sumar(5, 3)
print(f"La suma es: {resultado}")
```
