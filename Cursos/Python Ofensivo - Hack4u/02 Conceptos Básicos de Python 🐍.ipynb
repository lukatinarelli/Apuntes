# **🖥️ El Intérprete de Python**

El **intérprete** es el programa que **lee y ejecuta** el código Python. Es el “motor” que hace funcionar cualquier script o comando del lenguaje. Sin él, el código no es más que texto sin vida.

## **⚙️ ¿Qué hace exactamente?**

Su trabajo va más allá de solo "leer":

* **Ejecuta línea a línea:** Lee una instrucción y la ejecuta al momento. Esto es genial para encontrar errores (depuración) rápido.

* **Modo Interactivo (REPL):** Te permite abrir una terminal, escribir código y ver el resultado al instante (como hacemos en Jupyter).

* **Compila a Bytecode:** Antes de ejecutar, traduce tu código a un formato intermedio (`.pyc`) para ir más rápido.

* **Python Virtual Machine (PVM):** Ejecuta ese bytecode. Gracias a esta "máquina virtual", Python funciona igual en Windows, Linux o macOS sin que tú tengas que cambiar nada.

## **🚀 Por qué es útil**

* **Muy fácil de usar:** Ideal para aprender y probar ideas rápidamente sin configuraciones complejas.

* **Portable:** El mismo script que escribes en tu PC funcionará en el servidor Linux que ataques mañana.

* **Extensible:** Se puede ampliar con módulos en C/C++ para obtener más velocidad si hace falta.

# **🔖 Shebang y Convenios**

## **🐚 Shebang**

Es esa línea extraña que verás al inicio de muchos scripts de hacking. Indica al sistema operativo **qué intérprete debe usar** para ejecutar el archivo.

```bash
#!/usr/bin/env python3
```

* **¿Por qué usar `env`?** Es más portable. Si pones la ruta fija (`#!/usr/bin/python3`), el script fallará si el usuario instaló Python en otro sitio. Con `env`, el sistema busca dónde está instalado Python automáticamente.

* **En Hacking:** Es vital para que tus herramientas funcionen en cualquier distribución de Linux (Kali, Parrot, Ubuntu) sin dar errores.

## **📏 Convenios de codificación (PEP 8)**

Python tiene una "biblia" de estilo llamada **PEP 8**. Seguirla hace que tu código parezca escrito por un profesional y no por un aficionado.

### **1️⃣ Nombres (Naming Convention)**


Python es muy estricto con esto para mantener la coherencia:

| Elemento | Formato | Ejemplo |
| --- | --- | --- |
| **Variables y Funciones** | `snake_case` (minúsculas con guiones bajos) | `ip_victima`, `escanear_puerto` |
| **Constantes** | `UPPER_CASE` (mayúsculas) | `MAX_RETRIES`, `DEFAULT_PORT` |
| **Clases** | `PascalCase` (o camelCase) | `EscanerRed`, `ExploitManager` |

### **2️⃣ Indentación**

Aquí Python es único. No usa llaves `{ }` para definir bloques de código, usa **espacios**.

* **Regla:** 4 espacios por nivel.

* **Importante:** Si mezclas tabuladores y espacios, o fallas en la indentación, el programa dará error (`IndentationError`).

### **3️⃣ Estructura y Orden**

* **Longitud de línea:** Máximo 79 caracteres (para que se lea bien en terminales divididas).

* **Importaciones:** Siempre al principio del archivo y en este orden estricto:
    1. Librería estándar (ej: `os`, `sys`).

    2. Librerías de terceros (ej: `requests`, `scapy`).

    3. Tus propios módulos locales.

> 📌 **Nota:** Escribir código limpio (Clean Code) no es pillería. Cuando estés revisando tu propio exploit 6 meses después, agradecerás haber seguido estas normas.

# **📝 Variables y Tipos de Datos**

## **🏷️ Variables: Etiquetas**

En otros lenguajes, una variable es una "caja" donde guardas un dato. En Python, es más como una **etiqueta** que pegas a un objeto.

* Son **nombres que se asignan a datos**.

* **Tipado Dinámico:** No necesitas declarar el tipo (`int`, `char`). Python deduce qué es según el valor que le des.

````python
nombre = "Ana"     # Python ve comillas == String
edad = 25          # Python ve un número sin decimales == Integer
altura = 1.75      # Python ve un númeor con decimales == Float
````

## **🧱 Tipos de datos básicos**

Son los bloques de construcción fundamentales.

* **Cadenas (`str`)**: Texto. Son **inmutables** (no puedes cambiar una letra interna, tienes que crear una cadena nueva).

  * *Uso ofensivo:* Direcciones IP, contraseñas, payloads.

* **Números**:

  * **Enteros (`int`)**: Sin decimales.
    * *Uso ofensivo:* Números de puerto (ej: `80`, `443`).

  * **Flotantes (`float`)**: Con decimales.

* **Listas (`list`)**: Colecciones ordenadas y **mutables**.
  * *Uso ofensivo:* Lista de usuarios para fuerza bruta.

````python
# Una lista puede mezclar tipos (heterogénea)
mi_lista = [1, "hola", 3.14]
````

## **🔄 Iterando sobre datos (Bucles)**

Una de las tareas más comunes es recorrer una lista elemento por elemento.

````python
# 'elemento' es una variable temporal que toma el valor de cada item
for elemento in mi_lista:
    print(elemento)
````

> 📌 **Nota:** Estos son solo los tipos base. Más adelante veremos las estructuras potentes como **Diccionarios** (clave-valor) y **Tuplas**, vitales para manejar datos de redes y respuestas de servidores.

# **⚡ Operadores Básicos en Python**

Los operadores son los símbolos que le dicen al intérprete "haz algo con estos datos". Python es muy inteligente y el mismo símbolo puede hacer cosas distintas según el tipo de dato (esto se llama sobrecarga de operadores).

## **1️⃣ Operadores Aritméticos**

| Operador | Descripción | Ejemplo |
| --- | --- | --- |
| `+` | Suma / Concatenación | `2 + 3 → 5` |
| `-` | Resta | `5 - 2 → 3` |
| `*` | Multiplicación / Repetición | `2 * 3 → 6` |
| `/` | División (siempre devuelve float) | `5 / 2 → 2.5` |
| `//` | División entera (corta decimales) | `5 // 2 → 2` |
| `%` | Módulo (el resto de la división) | `5 % 2 → 1` (Útil para saber si es par/impar) |
| `**` | Exponente (Potencia) | `2 ** 3 → 8` |

## **2️⃣ Operadores con secuencias (Listas y Strings)**

Lo potente de Python es que sumar números es igual que sumar listas.

### **Cadenas (Strings)**

````python
"Hola " + "Mundo"      # Concatenación: "Hola Mundo"
````

````python
"Ha" * 3               # Repetición: "HaHaHa"
````

### **Listas**

````python
[1, 2] + [3, 4]        # Juntar listas: [1, 2, 3, 4]
````

````python
[0] * 5                # Inicializar lista: [0, 0, 0, 0, 0]
````

## **3️⃣ Funciones útiles para secuencias**

> 📌 **Nota:** En Python 3, `zip` y `map` son "lazy" (perezosos). Devuelven un objeto iterador. Para ver el resultado de golpe en el intérprete, los convertimos a `list()`.

````python
# zip: Une dos listas como una cremallera
list(zip(["Ana", "Luis"], [25, 30]))
````

````python
# map: Aplica una función a cada elemento
list(map(lambda x: x*2, [1, 2, 3]))
````

## **4️⃣ Conversión de tipos (TypeCast)**

A veces recibimos un número como texto ("10") y necesitamos operar con él.

```python
int("10") + 5    # 15 (Convierte string a entero)
str(5) + "00"    # "500" (Convierte entero a string)
float("3.14")    # 3.14 (Decimal)
```

# **📝 Formateo de Cadenas (String Formatting)**

Aquí hacemos una pausa importante sobre cómo mostramos los datos.

### **🛑 El intérprete vs. `print()`**

Hasta ahora, en *Jupyter/Colab*, si escribes `"Hola"`, ves el resultado `'Hola'` con comillas. Eso es la **representación interna** del objeto.
Pero cuando creamos programas reales para usuarios, queremos mostrar texto limpio. Ahí entra `print()`.

* **Intérprete:** Muestra el dato crudo (útil para debug).

* **print():** Muestra el dato formateado para humanos.

## **1️⃣ F-Strings (Moderno y Recomendado) 🔥**

Introducidas en Python 3.6, son la forma más rápida y legible. Solo pones una `f` antes de las comillas y usas llaves `{}`.

````python
nombre = "Marta"
edad = 28

# Fíjate que aquí SÍ usamos print para ver el resultado final limpio
print(f"Hola {nombre}, tienes {edad} años")
````

````python
# Admiten expresiones matemáticas dentro
print(f"En 5 años tendrás {edad + 5}")
````

## **2️⃣ Método `.format()` (La forma anterior)**

Útil si usas versiones antiguas de Python (< 3.6), aunque es más verboso.

````python
print("Hola {}, tienes {} años".format("Luis", 30))
````

````python
print("Pi vale {:.2f}".format(3.14159))  # .2f limita a 2 decimales
````

## **3️⃣ Operador `%` (Legacy / Estilo C)**

Es la forma antigua. Aún se ve en exploits viejos, pero evita usarla en código nuevo.

````python
print("Hola %s, tienes %d años" % ("Ana", 25))
````

# **🧭 Control de Flujo**

Hasta ahora, nuestro código se ejecutaba línea por línea, de arriba a abajo. Con el **Control de Flujo**, podemos hacer que el programa tome decisiones o repita tareas.

## **1️⃣ Condicionales (`if`, `elif`, `else`)**

Permiten ejecutar bloques de código **solo si se cumple una condición**. Es la lógica básica de cualquier toma de decisiones.

if-else-flowchart-programming.png

> 📌 **Nota:** recuerda que la indentación (4 espacios o un tab) define qué líneas pertenecen al if.

````python
edad = 20  # ⬅️ Modifica este valor para ver qué pasa

if edad > 18:
    print("Acceso concedido: Eres mayor de edad")
elif edad == 18:
    print("Acceso concedido: Justo en el límite")
else:
    print("Acceso denegado: Eres menor")
````

## **2️⃣ Bucles (Loops)**

Los bucles sirven para **repetir** un bloque de código muchas veces sin tener que escribirlo una y otra vez.

loop-diagram-programming.png

### **🔁 Bucle `for`**

Se usa cuando sabemos **cuántas veces** queremos iterar o tenemos una lista de cosas que recorrer.

````python
# Recorrer una lista
colores = ["Rojo", "Verde", "Azul"]

for color in colores:
    print(f"Color actual: {color}")
````

````python
# Usando range() para contar. range(3) genera: 0, 1, 2
for i in range(3):
    print(f"Intento número {i}")
````

### **🔄 Bucle `while`**

Se usa cuando **no sabemos cuándo acabará**, solo sabemos que debe seguir ejecutándose **mientras** una condición sea verdadera.

````python
contador = 0

while contador < 3:
    print(f"El contador vale {contador}")
    contador += 1  # ¡Importante aumentar para no crear un bucle infinito!
````

## **3️⃣ Control del flujo interno**

A veces necesitamos alterar el comportamiento normal de un bucle.

* **`break`**: Rompe el bucle **de golpe** y sale.
  * *Uso ofensivo:* Si estás haciendo fuerza bruta y encuentras la contraseña, haces un `break` para dejar de probar.


* **`continue`**: Salta lo que queda de código en **esa vuelta** y pasa a la siguiente.
  * *Uso ofensivo:* Si escaneas una IP y no responde al ping, haces `continue` para pasar a la siguiente sin intentar escanear sus puertos.


* **`pass`**: No hace nada. Es un relleno (placeholder).
  * *Uso:* Cuando estás diseñando la estructura de un script pero aún no quieres escribir el código de una parte.

````python
for x in range(10):
    if x == 2:
        continue  # Salta el 2 (no lo imprime)
    if x == 5:
        break     # Al llegar al 5, se acaba todo el bucle

    print(x)
````

# **📌 Funciones y Ámbito (Scope)**

## **1️⃣ Funciones: El principio DRY**

**DRY** = *Don't Repeat Yourself*. Si copias y pegas código, estás haciéndolo mal.
Las funciones nos permiten encapsular lógica para usarla cuantas veces queramos.

````python
# Definición
def saludar(nombre):
    """
    Docstring: Esta función recibe un nombre y devuelve un saludo.
    """
    mensaje = f"Hola, {nombre}"

    return mensaje

# Llamada a la función
print(saludar("Mr. Robot"))
````

> 💡 **Tip:** El `return` es crucial. Si usas `print()` dentro de la función, la función muestra algo pero devuelve `None`. Si quieres guardar el resultado en una variable, usa `return`.

## **2️⃣ Scope (Ámbito de las variables)**

El **Scope** decide desde dónde puedes "ver" o usar una variable. Piensa en ello como cristales tintados:

1. **Local:** Variables creadas *dentro* de una función. Solo existen allí. Si intentas usarlas fuera, darán error.

2. **Global:** Variables creadas en el cuerpo principal. Se pueden leer desde cualquier sitio.

### **Scope Local**

````python
def funcion_local():
    secreto = "Soy una variable local"

    print(secreto)

funcion_local()  # Funciona
````

````python
# Si intentas esto, dará error porque 'secreto' murió al acabar la función
print(secreto)  # NameError
````

### **Scope Global y la palabra `global`**

Por seguridad, Python no te deja **modificar** una variable global desde dentro de una función a menos que le des permiso explícito.

````python
x = 5  # Variable Global

def intentar_cambiar():
    x = 10  # Python crea una NUEVA variable local llamada x, no toca la de fuera
    print(f"Dentro: {x}")

intentar_cambiar()
print(f"Fuera tras primer intento: {x}")  # Sigue siendo 5
````

````python
x = 5

def cambiar_de_verdad():
    global x  # ⚠️ Permiso para modificar la variable de fuera
    x = 20
    print(f"Dentro (con global): {x}")

cambiar_de_verdad()
print(f"Fuera tras segundo intento: {x}") # Ahora es 20
````

## **3️⃣ Buenas prácticas**

* **Funciones cortas:** Si tu función ocupa más de una pantalla, divídela.

* **Nombres claros:** `calcular_hash()` mejor que `func1()`.

* **Evitar `global`:** Hacen que el código sea impredecible y difícil de depurar. Es mejor pasar los valores como argumentos y usar `return`.

# **📌 Funciones Lambda (Anónimas)**

## **¿Qué son?**

Son funciones "de usar y tirar". Se escriben en una sola línea y no tienen nombre. Son muy comunes en programación funcional y análisis de datos.

* **Sintaxis:** `lambda argumentos: expresión`

````python
# Versión clásica: El doble de un número
def doble(x):
    return x * 2

# Versión Lambda equivalente
lambda x: x * 2
````

## **Usos típicos**

Brillan cuando necesitas pasar una función pequeña como argumento a otra función (como `map`, `filter` o `sorted`).

````python
numeros = [1, 2, 3, 4, 5, 6]

# MAP: Aplica la lambda a cada elemento
dobles = list(map(lambda x: x * 2, numeros))

print(f"Dobles: {dobles}")
````

````python
numeros = [1, 2, 3, 4, 5, 6]

# FILTER: Filtra los que cumplan la condición (True)
pares = list(filter(lambda x: x % 2 == 0, numeros))

print(f"Pares: {pares}")
````

## **Limitaciones**

* Solo pueden contener una **única expresión**.

* No pueden tener bloques complejos (`if/else` largos, bucles `for`, etc.).

* Si es compleja, usa una función normal (`def`).

# **⚠️ Excepciones y Manejo de Errores**

## **1️⃣ ¿Qué es una Excepción?**

Es un evento que ocurre durante la ejecución del programa y **rompe el flujo normal** (un error).
Si no las manejas, el script se detiene (crashea) y muestra un mensaje en rojo.

> 👨‍💻 **En Hacking:** Imagina un script escaneando 1000 IPs. Si la IP #5 da error y no tienes manejo de excepciones, el script se para y pierdes el progreso de las otras 995.

## **2️⃣ La estructura `try` ... `except`**

Es la red de seguridad. Le dices a Python: *"Intenta hacer esto, y si falla, no te rompas, haz esto otro"*.

````python
try:
    # Código que podría fallar
    numero = int("hola")

except ValueError:
    # Código que se ejecuta SI hay un error específico
    print("❌ Error: No puedes convertir texto a número.")
````

## **3️⃣ El flujo completo: `else` y `finally`**

Podemos hacer la lógica más compleja:

* **`else`**: Se ejecuta solo si **NO** hubo errores en el `try`.

* **`finally`**: Se ejecuta **SIEMPRE**, haya error o no (ideal para limpiezas).

````python
try:
    archivo = open("datos.txt")

except FileNotFoundError:
    print("❌ Error: El archivo no existe.")

else:
    print("✅ Archivo abierto correctamente (Aquí leemos los datos).")
    archivo.close()

finally:
    print("🏁 Operación de apertura finalizada (Esto sale siempre).")
````

## **4️⃣ Lanzar excepciones (`raise`)**

A veces quieres provocar un error tú mismo. Por ejemplo, si una función recibe un dato que no te gusta (validación).

````python
divisor = 0

if divisor == 0:
    # Creamos nuestro propio error para detener el programa controladamente
    raise ZeroDivisionError("⚠️ Alerta: No se permite dividir entre cero")
````

## **5️⃣ Excepciones comunes (Cheat Sheet)**

| Excepción | Cuándo ocurre |
| --- | --- |
| `ValueError` | Tipo correcto pero valor inapropiado (ej: `int("texto")`). |
| `TypeError` | Operación en tipos incorrectos (ej: `"3" + 5`). |
| `IndexError` | Intentar acceder a una posición que no existe en una lista. |
| `KeyError` | Buscar una clave que no existe en un diccionario. |
| `ZeroDivisionError` | Dividir por cero. |
| `FileNotFoundError` | Intentar abrir un archivo inexistente. |

## **6️⃣ Buenas prácticas**

* ✔ **Sé específico:** Usa `except ValueError` en lugar de un `except:` genérico. Si capturas todo, ocultarás bugs reales.

* ✔ **Mensajes claros:** Que el usuario sepa qué ha fallado.

* ✔ **Usa `finally` para limpiar:** Cerrar conexiones a bases de datos, sockets de red o ficheros abiertos.
