# **Colecciones y Estructuras de Datos 📚**

En Python, cuando programamos, a menudo necesitamos guardar muchos datos a la vez. En lugar de crear mil variables (`dato1`, `dato2`, `dato3`...), usamos **colecciones**.

Piensa en ellas como diferentes tipos de "contenedores" para organizar tu información.

### **Las 4 principales son:**

1. **📦 Listas (Lists):** Para guardar datos en orden. Son flexibles (puedes cambiar lo que hay dentro).

2. **🔒 Tuplas (Tuples):** Igual que las listas, pero "cerradas". Una vez creadas, no se pueden modificar.

3. **🔷 Conjuntos (Sets):** Para guardar elementos únicos. No permiten duplicados.

4. **📘 Diccionarios (Dictionaries):** Para guardar información conectada (Clave -> Valor), como una agenda.

## **1. 📦 Listas (Lists)**

Las listas son la estructura más usada. Piensa en ellas como una fila de elementos ordenados.

Son **heterogéneas** (puedes mezclar números y texto) y **mutables** (puedes cambiar su contenido).

```python
# Crear una lista y mostrarla
mi_lista = [10, "hola", 3.5, True]
print(mi_lista)
```

### **🔢 Acceder a elementos**

Usamos el índice (posición). Recuerda que **empieza en 0**.

```python
frutas = ["Manzana", "Banana", "Cereza"]

# Acceder al primero
print(frutas[0])

# Acceder al último (índice negativo)
print(frutas[-1])
```

### **📝 Modificar elementos**

Como las listas son mutables, podemos cambiar lo que hay dentro.

```python
# Cambiamos el segundo elemento (índice 1)
frutas[1] = "Pera"
print(frutas)
```

### **🛠️ Añadir y Eliminar**

Podemos hacer la lista más grande o más pequeña.

```python
colores = ["Rojo", "Verde"]

# Añadir al final
colores.append("Azul")
print(colores)

# Eliminar por nombre
colores.remove("Rojo")
print(colores)
```

### **✂️ Slicing (Cortar listas)**

Para quedarnos solo con un trozo. Formato: `[inicio : fin]` (el final no se incluye).

```python
numeros = [0, 1, 2, 3, 4, 5, 6]

# Del índice 2 al 5 (sin incluir el 5)
print(numeros[2:5])
```

---

## **2. 🔒 Tuplas (Tuples)**

Las tuplas son casi idénticas a las listas, pero tienen una gran diferencia: son **INMUTABLES**.

Una vez creas una tupla, **no puedes cambiar, añadir ni borrar nada**. Son datos fijos.
Se definen con paréntesis `( )`.

```python
# Crear una tupla
mi_tupla = (10, 20, 30)
print(mi_tupla)
```

### **👁️ Leer datos**

Se hace exactamente igual que en las listas, usando el índice `[ ]`.

```python
# Ver el primer elemento
print(mi_tupla[0])
```

### **🚫 La regla de oro (Inmutabilidad)**

Aquí es donde fallan si intentas tratarlas como listas. No puedes modificar sus valores.

```python
# ESTO DARÍA ERROR
print(mi_tupla[0] = 99)
```

### **¿Cuándo usarlas?**

Úsalas para datos que no deben cambiar nunca durante la ejecución del programa, como coordenadas GPS, días de la semana o configuraciones de pantalla.

```python
# Ejemplo: Coordenadas (Latitud, Longitud)
punto_gps = (40.416, -3.703)
print(punto_gps)
```

---

## **3. 🔷 Conjuntos (Sets)**

Son colecciones desordenadas (no tienen índice 0, 1, 2...) y su regla de oro es: **NO admiten duplicados**.

Si intentas meter el mismo dato dos veces, Python lo elimina automáticamente. Se definen con llaves `{ }`.

```python
# Creamos un conjunto con números repetidos
mi_conjunto = {1, 2, 3, 3, 3, 4, 4}

# Al mostrarlo, los repetidos han desaparecido
print(mi_conjunto)
```

### **🔍 Búsqueda rápida**

Los sets son buenísimos para preguntar: "¿Está este dato aquí?". Usamos la palabra clave `in`.

```python
# ¿Está el número 2 en el conjunto?
2 in mi_conjunto
```

### **🛠️ Añadir y Eliminar**

Como son mutables, podemos meter y sacar cosas, pero no usar índices (`mi_conjunto[0]` daría error).

```python
# Añadir un elemento
mi_conjunto.add(5)
print(mi_conjunto)
```

```python
# Eliminar un elemento
mi_conjunto.remove(1)
print(mi_conjunto)
```

---

## **4. 📘 Diccionarios (Dictionaries)**

Funcionan con parejas **Clave: Valor**. Son como una agenda telefónica real: buscas un nombre (clave) y obtienes su número (valor).

Se definen con llaves `{ }`, pero a diferencia de los sets, aquí los datos van en parejas.

```python
# Definimos una persona con sus datos
persona = {
    "nombre": "Carlos",
    "edad": 28,
    "ciudad": "Madrid"
}

print(persona)
```

### **🔑 Acceder a los datos**

Aquí no usamos índices numéricos (0, 1...). Usamos el **nombre de la clave**.

```python
# ¿Cómo se llama?
print(persona["nombre"])

# ¿Qué edad tiene?
print(persona["edad"])
```

### **📝 Modificar y Añadir**

Podemos cambiar valores existentes o inventarnos claves nuevas sobre la marcha.

```python
# Modificar: Cumpleaños (cambiamos 28 por 29)
persona["edad"] = 29
print(persona)
```

```python
# Añadir: Nueva clave 'profesion'
persona["profesion"] = "Hacker"
print(persona)
```

### **📋 Ver solo las claves**

A veces solo queremos saber qué tipo de información tenemos guardada.

```python
print(persona.keys())
```
