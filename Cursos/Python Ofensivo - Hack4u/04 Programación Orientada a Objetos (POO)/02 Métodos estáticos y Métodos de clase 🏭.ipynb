# **🛠️ Métodos de Clase y Métodos Estáticos en Python (POO)**

En la POO, a veces necesitamos funciones que pertenezcan a la clase, pero que no necesariamente necesiten acceder a los datos de un objeto concreto (como su nombre o edad). Para eso existen estos dos decoradores.

## **🏷️ 1. Métodos de Clase (`@classmethod`)**

La clave aquí es entender que **una clase en Python es también un objeto**.
Un método de clase no trabaja sobre una instancia individual (un `hacker1`), sino sobre el **molde** (`Persona`).

* **Decorador:** `@classmethod`

* **Primer argumento:** `cls` (en lugar de `self`).

  * `self` apunta a "este objeto".

  * `cls` apunta a "la clase constructora".

### **¿Para qué sirven? (El patrón Factory)**

Su uso principal en Hacking y desarrollo avanzado es actuar como **Constructores Alternativos (Factories)**.
Imagina que recibes datos sucios (un string, un JSON, un archivo) y quieres crear un objeto limpio directamente desde ahí.

#### Ejemplo mejorado:

```python
class Persona:
    especie = "Humano"  # Atributo de clase

    def __init__(self, nombre):
        self.nombre = nombre

    # 1. USO: Modificar el estado global de la clase
    @classmethod
    def cambiar_especie(cls, nueva_especie):
        cls.especie = nueva_especie
        # Al usar 'cls', cambiamos la variable para TODAS las instancias futuras y presentes

    # 2. USO: Factory Method (Constructor alternativo)
    @classmethod
    def crear_desde_texto_sucio(cls, texto):
        # Imaginemos que recibimos "  Usuario:Lucía-Admin  "
        nombre_limpio = texto.strip().split("-")[0].replace("Usuario:", "")

        # IMPORTANTE: Usamos 'cls(nombre)' en vez de 'Persona(nombre)'
        # Esto permite que si alguien hereda de esta clase, el método siga funcionando.
        return cls(nombre_limpio)

# --- USO ---

# 1. Usando el Factory
# No llamamos a Persona(), llamamos al método que nos devuelve una Persona ya configurada
p = Persona.crear_desde_texto_sucio("  Usuario:Lucía-Admin  ")
print(p.nombre)
```

```python
# 2. Cambiando la especie global
Persona.cambiar_especie("Cyborg")
print(p.especie)
print(Persona.especie)
```

> 💡 **¿Por qué `cls` y no el nombre de la clase?**
>
> Podrías escribir `Persona(nombre_limpio)` dentro del método, pero usar `cls(nombre_limpio)` es mejor práctica.
> Si mañana creas una clase hija `class Hacker(Persona)`, el método `crear_desde_texto_sucio` usará automáticamente la clase `Hacker` gracias a `cls`.

## **🔧 2. Métodos Estáticos (`@staticmethod`)**

Estos métodos son, en esencia, **funciones normales** que decidimos guardar dentro de una clase por orden y limpieza.

* **Sintaxis:** Llevan el decorador `@staticmethod`.

* **Argumentos:** **NO** reciben ni `self` (instancia) ni `cls` (clase).

* **Independencia:** No pueden acceder a los datos del objeto ni modificar la clase. Son 100% independientes.

### **¿Por qué usarlos entonces?**

Si son funciones normales, ¿por qué no dejarlas fuera de la clase?
Por **Organización (Namespacing)**. Si tienes una clase `Matematicas`, tiene sentido que la función de sumar esté dentro de ella (`Matematicas.sumar`), en lugar de tener una función `sumar` flotando suelta en tu código que podría confundirse con otras sumas.

#### Ejemplo simple:

```python
class Calculadora:
    # Método estático: No necesita 'self' porque sumar no depende de nada guardado
    @staticmethod
    def sumar(a, b):
        return a + b

# No hace falta crear un objeto
print(Calculadora.sumar(5, 5))
```

> 📌 **Nota:** Un método estático es como una herramienta que llevas en la caja, pero que funciona igual fuera de ella. Está dentro solo para que no se pierda.

## **🧠 3. Diferencias clave y Cuándo usar cuál**

Esta es la parte más importante. Muchos programadores confunden `@classmethod` con `@staticmethod`. Aquí tienes la guía definitiva.

### **Tabla Comparativa**

| Tipo de Método | Decorador | 1er Argumento | ¿A qué tiene acceso? | Uso principal |
| --- | --- | --- | --- | --- |
| **Instancia** | (nada) | `self` | Al objeto específico | Modificar datos del objeto (`self.nombre`). |
| **Clase** | `@classmethod` | `cls` | A la clase entera | Crear objetos (Factories) o cambiar config global. |
| **Estático** | `@staticmethod` | Ninguno | A nada | Funciones de ayuda o utilidades puras. |

### **Diagrama de decisión (Mental)**

Cuando escribas una función dentro de una clase, hazte estas preguntas en orden:

1. **¿Necesito acceder a variables del objeto (`self.algo`)?**
    * ✅ SÍ → Es un **Método de Instancia** (lo normal).
    * ❌ NO → Sigue a la pregunta 2.


2. **¿Necesito crear una nueva instancia de la clase o acceder a atributos globales (`cls.algo`)?**
    * ✅ SÍ → Es un **Método de Clase** (`@classmethod`).
    * ❌ NO → Sigue a la pregunta 3.


3. **¿Es solo una función auxiliar que no toca ni la clase ni el objeto?**
    * ✅ SÍ → Es un **Método Estático** (`@staticmethod`).

# **📌 En Resumen**

* **`@classmethod`**: Piensa en **"Fábricas"**. Sirven para construir objetos de formas alternativas.
* **`@staticmethod`**: Piensa en **"Utilidades"**. Son funciones aisladas agrupadas por lógica.
* Ambos ayudan a que tu código no sea un caos de funciones sueltas y esté todo bien encapsulado en sus clases correspondientes.
