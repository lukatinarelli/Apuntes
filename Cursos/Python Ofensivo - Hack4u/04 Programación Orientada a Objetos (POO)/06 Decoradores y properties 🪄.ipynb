# **🧩 Decoradores y Properties en Python**

Los **decoradores** y **properties** son herramientas avanzadas que nos permiten escribir código más limpio, reutilizable y "mágico". Nos permiten interceptar la ejecución de funciones o el acceso a variables para añadir lógica extra sin ensuciar el código principal.

## **🛠️ 1. Decoradores: Modificar funciones**

Un **decorador** no es más que una función que **recibe otra función**, le añade cosas (la "decora") y devuelve una nueva función mejorada. Piensa en ello como ponerle una carcasa a un móvil: el móvil es el mismo, pero ahora es resistente al agua.

#### Ejemplo básico: El Logger

Imagina que quieres saber cuándo se ejecuta una función y con qué argumentos, muy útil para *debugging* de exploits.

```python
# 1. El Decorador (La carcasa)
def debug(func):

    # 2. El Wrapper (La lógica extra que envuelve a la original)
    def wrapper(*args, **kwargs):
        print(f"🕵️ [DEBUG] Ejecutando: {func.__name__}")
        print(f"   ∟ Args: {args}")
        print(f"   ∟ Kwargs: {kwargs}\n")

        # 3. Llamada original (Lo que hace la función realmente)
        resultado = func(*args, **kwargs)

        print(f"✅ Terminada: {func.__name__}")
        return resultado

    return wrapper

# Usamos el decorador con la arroba @
@debug
def atacar_objetivo(ip, puerto=80):
    print(f"🚀 Lanzando payload a {ip}:{puerto}")

# Al llamar a la función, se ejecuta el wrapper
atacar_objetivo("192.168.1.5", puerto=443)
```

### **Ventajas Clave**

* **No tocas el código original:** La función `atacar_objetivo` sigue limpia.

* **Reutilización:** Puedes poner `@debug` a 50 funciones distintas sin escribir ni una línea más de código.

* **Validación:** Puedes usar decoradores para comprobar si un usuario es `admin` antes de ejecutar una función crítica.

## **🏷️ 2. Properties: Control de acceso a atributos**

Las **properties** son la evolución de los *Getters* y *Setters*. Nos permiten interceptar el acceso a una variable para validar datos, pero manteniendo una sintaxis limpia.

Para quien usa tu clase, parece que está tocando una variable normal (`p.nombre = "Ana"`), pero en realidad está pasando por tu filtro de seguridad.

#### Ejemplo con `@property`:

```python
class Persona:
    def __init__(self, nombre):
        # Usamos el setter desde el principio para validar el nombre inicial
        self.nombre = nombre

    # 1. GETTER: Se llama cuando pides p.nombre
    @property
    def nombre(self):
        """Devuelve el nombre protegido"""
        return self._nombre

    # 2. SETTER: Se llama cuando haces p.nombre = "Valor"
    @nombre.setter
    def nombre(self, valor):
        """Valida el dato antes de guardarlo"""
        if not valor:
            raise ValueError("❌ Error: El nombre no puede estar vacío")
        print(f"✅ Guardando nombre: {valor}")
        self._nombre = valor

    # 3. DELETER: Se llama cuando haces del p.nombre
    @nombre.deleter
    def nombre(self):
        print("🗑️ Eliminando nombre...")
        del self._nombre

# USO
p = Persona("Lucía") # Pasa por el setter
print(p.nombre)      # Pasa por el getter -> "Lucía"

p.nombre = "Ana"     # Pasa por el setter -> "Guardando nombre: Ana"

p.nombre = ""      # ¡Salta el error ValueError! No deja guardar basura.
```

### **Ventajas Clave**

1. **Validación Ofensiva:** Puedes asegurar que un puerto sea int (`1-65535`) o que una IP tenga formato válido antes de asignarla a tu objeto.

2. **Compatibilidad:** Puedes empezar con un atributo público normal. Si en el futuro necesitas añadir validación, lo conviertes en una *property* y **no rompes el código** de quien ya usaba tu clase.

## **💡 3. Uso combinado y Resumen**

Los decoradores y properties son la base de la metaprogramación en Python.

* **Decoradores (`@func`):** Envuelven funciones para añadir logs, medir tiempos o verificar permisos.

* **Properties (`@property`):** Envuelven atributos para añadir lógica de acceso (Get/Set) sin cambiar la sintaxis.
