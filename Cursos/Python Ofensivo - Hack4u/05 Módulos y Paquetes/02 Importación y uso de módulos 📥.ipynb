# **📥 Importación y uso de módulos**

Una vez que hemos dividido nuestro código en varios archivos (módulos), necesitamos una forma de unirlos cuando el programa se ejecuta. Para eso sirve la sentencia `import`.

## **1️⃣ La sentencia `import` (Importación completa)**

Es la forma más básica. Le dices a Python: "Trae el archivo entero".
Para usar cualquier función dentro de ese módulo, debes usar la sintaxis del punto: `modulo.funcion()`.

#### Ejemplo con la librería estándar `math`:

```python
import math

# Debemos usar el prefijo 'math.' para acceder a sus herramientas
raiz = math.sqrt(25)

print(f"La raíz cuadrada de 25 es: {raiz}")
print(f"El valor de PI es: {math.pi}")
```

> 📌 **Nota:** Esto es muy ordenado porque sabes exactamente de dónde viene cada función. Si lees `math.sqrt`, sabes que es del módulo `math`.

## **2️⃣ Alias con `as` (Renombrar módulos)**

A veces los nombres de los módulos son muy largos o simplemente queremos escribir menos. Podemos ponerles un mote (alias) usando `as`.

Es muy común en Data Science (`import pandas as pd`, `import numpy as np`).

```python
import math as m

# Ahora usamos 'm' en lugar de 'math'
print(m.sqrt(100))
```

## **3️⃣ Importación selectiva (`from ... import ...`)**

Si solo necesitas una herramienta específica (por ejemplo, solo el martillo) y no quieres cargar con toda la caja de herramientas, usas `from`.

**Ventaja:** No necesitas escribir el prefijo del módulo (`math.`) cada vez.
**Desventaja:** Pierdes un poco de contexto (no es tan obvio de dónde viene la función).

```python
from math import sqrt, pi

# ¡Fíjate! Ya no usamos 'math.sqrt', la usamos directamente
resultado = sqrt(49)
print(resultado)
print(pi)
```

## **⚠️ 4️⃣ La importación peligrosa (`from ... import *`)**

Puedes importar **todo** el contenido de un módulo directamente al espacio de nombres principal usando el asterisco `*`.

from math import * # Esto importa sqrt, pi, sin, cos, tan... TODO de golpe.


> ⚠️ **¿Por qué NO se recomienda?**
>
> Esto causa **Contaminación del Espacio de Nombres**.
> Imagina que tú tienes una variable llamada `pi = 3`. Si haces `from math import *`, la `pi` de matemáticas sobrescribirá tu variable sin avisar. Es mejor ser explícito.

## **5️⃣ ¿Cómo encuentra Python los módulos? (`sys.path`)**

Cuando escribes `import hack_tools`, ¿cómo sabe Python dónde buscar ese archivo? ¿Busca en mis documentos? ¿En el escritorio?

Python sigue una ruta de búsqueda estricta almacenada en una lista llamada `sys.path`. Busca en este orden:

1. **Directorio actual:** La carpeta donde estás ejecutando el script.
2. **PYTHONPATH:** Una variable de entorno (si la has configurado).
3. **Librería Estándar:** Donde se instaló Python.
4. **Librerías de terceros:** (La carpeta `site-packages`).

```python
import sys

# Vamos a ver dónde está buscando Python en tu máquina
for ruta in sys.path:
    print(ruta)
```

> 💡 **Debug Hack**
>
> Si alguna vez te sale el error `ModuleNotFoundError: No module named 'x'`, ejecuta el código de arriba. Seguramente tu archivo no está en ninguna de esas carpetas.

## **6️⃣ Tipos de Módulos**

En tu carrera verás tres tipos de módulos:

### **A. Módulos de la Biblioteca Estándar (Built-in)**

Vienen instalados con Python. Son "baterías incluidas".

* `os`: Interactuar con el sistema operativo.
* `sys`: Interactuar con el intérprete.
* `re`: Expresiones regulares.
* `socket`: Conexiones de red.

### **B. Módulos de Terceros**

Hechos por la comunidad. Se instalan con `pip` (el gestor de paquetes).

* `requests`: Para peticiones HTTP.
* `scapy`: Para manipular paquetes de red.

### **C. Módulos Propios**

Los archivos `.py` que tú creas (como vimos en la clase anterior).
