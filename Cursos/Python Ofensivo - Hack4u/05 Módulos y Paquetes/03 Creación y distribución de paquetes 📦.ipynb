# **📦 Creación y distribución de paquetes**

Ya sabemos crear **módulos** (archivos sueltos). Pero, ¿qué pasa cuando nuestro proyecto de hacking se vuelve enorme? ¿Qué pasa si tenemos 50 archivos de utilidades? No podemos tenerlos todos tirados en la misma carpeta.

Aquí entran los **Paquetes**.

## **1️⃣ ¿Qué es un Paquete?**

En palabras simples: **Un paquete es una carpeta que contiene módulos.**

Pero no cualquier carpeta sirve. Para que Python entienda que una carpeta es importable (es decir, que puedes hacer `import mi_carpeta`), debe contener un archivo especial (y a veces mágico) llamado `__init__.py`.

### **Estructura típica**

Imagina que queremos crear una suite de herramientas llamada `hacktools`. La estructura sería así:

```text
hacktools/            <-- La carpeta (El Paquete)
    ├── __init__.py   <-- ¡OBLIGATORIO! (Indica que es un paquete)
    ├── red.py        <-- Módulo de red
    └── sistema.py    <-- Módulo de sistema

```

## **2️⃣ Manos a la obra: Creando nuestro paquete**

Vamos a usar código para crear esta estructura de carpetas y archivos ahora mismo en tu entorno.

### **Paso A: Crear la carpeta y el `__init__.py`**

````python
import os

# 1. Creamos la carpeta principal 'hacktools'
if not os.path.exists('hacktools'):
    os.makedirs('hacktools')

# 2. Creamos el archivo __init__.py (puede estar vacío)
# El comando %%writefile no funciona dentro de un if, así que usamos open()
with open('hacktools/__init__.py', 'w') as f:
    f.write("# Este archivo hace que 'hacktools' sea un paquete\n")
    f.write("print('📦 Cargando el paquete hacktools...')")
````

> 💡 **El archivo `__init__.py`:**
>
> Normalmente se deja vacío, pero todo código que escribas dentro se ejecutará **automáticamente** en el momento en que alguien haga `import hacktools`. Es útil para inicializar configuraciones.

### **Paso B: Crear los módulos internos**

Ahora llenemos el paquete con un par de herramientas.

````python
%%writefile hacktools/red.py

def escanear_ip(ip):
    return f"🔍 Escaneando la IP {ip}..."
````

````python
%%writefile hacktools/sistema.py

def obtener_os():
    return "💻 Sistema Operativo detectado: Linux (Kali)"
````

## **3️⃣ Importando desde un Paquete**

Ahora que la estructura existe, Python nos permite navegar por las carpetas usando **puntos**.

**Sintaxis:** `from paquete.modulo import funcion`

````python
# Fíjate que al ejecutar esto, saldrá el print que pusimos en __init__.py
from hacktools.red import escanear_ip
from hacktools.sistema import obtener_os

print(escanear_ip("192.168.1.10"))
print(obtener_os())
````

## **4️⃣ Distribución: Compartir tu código (`setup.py`)**

Si quisieras enviarle esta herramienta a un amigo o subirla a internet para que cualquiera la instale con `pip install hacktools`, necesitas un archivo de configuración llamado `setup.py` fuera de la carpeta del paquete.

Este archivo es como el DNI de tu proyecto.

````python
%%writefile setup.py
from setuptools import setup, find_packages

setup(
    name="hacktools_pro",        # Nombre para instalar (pip install hacktools_pro)
    version="1.0",
    description="Mis herramientas de hacking ofensivo",
    author="Tu Nombre",
    packages=find_packages()     # Busca automáticamente carpetas con __init__.py
)
````

### **El repositorio PyPI (Python Package Index)**

Cuando haces `pip install pandas` o `pip install requests`, tu ordenador se conecta a **PyPI** (el almacén mundial de librerías de Python).

1. Creas tu paquete.
2. Creas el `setup.py`.
3. Lo subes a PyPI.
4. ¡Cualquier persona en el mundo puede usar tu código!

> 📌 **Instalación Local**
>
> No hace falta subirlo a internet para probarlo. Si estás en la carpeta donde está el `setup.py`, puedes ejecutar en tu terminal:
> `pip install .`
> Esto instalará tu paquete en tu ordenador como si fuera una librería oficial.
