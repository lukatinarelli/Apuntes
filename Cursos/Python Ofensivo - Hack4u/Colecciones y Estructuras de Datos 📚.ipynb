# **Colecciones y Estructuras de Datos 📚**

En Python, cuando programamos, a menudo necesitamos guardar muchos datos a la vez. En lugar de crear mil variables (`dato1`, `dato2`, `dato3`...), usamos **colecciones**.

Piensa en ellas como diferentes tipos de "contenedores" para organizar tu información.

### **Las 4 principales son:**

1. **📦 Listas (Lists):** Para guardar datos en orden. Son flexibles (puedes cambiar lo que hay dentro).
2. **🔒 Tuplas (Tuples):** Igual que las listas, pero "cerradas". Una vez creadas, no se pueden modificar.
3. **🔷 Conjuntos (Sets):** Para guardar elementos únicos. No permiten duplicados.
4. **📘 Diccionarios (Dictionaries):** Para guardar información conectada (Clave -> Valor), como una agenda.

## **1. 📦 Listas (Lists)**

Las listas son la estructura más usada. Piensa en ellas como una fila de elementos ordenados.

Son **heterogéneas** (puedes mezclar números y texto) y **mutables** (puedes cambiar su contenido).

```python
# Crear una lista es fácil, usamos corchetes [ ]
mi_lista = [10, "hola", 3.5, True]

# Ver la lista completa
print(mi_lista)
```

### **🔢 Acceder y Modificar elementos**

Cada elemento tiene una posición (índice). **¡Ojo! En programación empezamos a contar desde 0.**

Como son mutables, podemos reescribir cualquier posición.

```python
frutas = ["Manzana", "Banana", "Cereza"]

# 1. Acceder (Leer)
print(frutas[0])  # Manzana
print(frutas[-1]) # Cereza (índice negativo va al final)

# 2. Modificar (Escribir)
frutas[1] = "Pera" # Cambiamos Banana por Pera

print(frutas)
```

### **🛠️ Añadir y Eliminar**

Las listas son dinámicas. Puedes hacerlas crecer o encoger según necesites.

```python
colores = ["Rojo", "Verde"]

# AÑADIR: .append() lo pone al final
colores.append("Azul")

# INSERTAR: .insert() lo pone donde tú digas
colores.insert(0, "Amarillo") # Lo pone el primero

# ELIMINAR: .remove() busca el elemento y lo borra
colores.remove("Rojo")

print(colores)
```

### **✂️ Slicing (Cortar listas)**

A veces solo quieres un trozo de la lista, no toda. Para eso usamos el "slicing" (rebanar).

La regla es: `[inicio : fin]` (el final no se incluye).

```python
numeros = [0, 1, 2, 3, 4, 5, 6]

# Cortar desde el índice 2 hasta el 5 (sin incluir el 5)
sub_lista = numeros[2:5] 

print(sub_lista) # Resultado: [2, 3, 4]
```
