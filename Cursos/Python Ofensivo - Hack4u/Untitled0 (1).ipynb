# **🐍 Historia y filosofía de Python**

Python no es solo un lenguaje, es una filosofía de desarrollo. Fue creado por **Guido van Rossum** a finales de los 80 (lanzado en 1991) con un objetivo claro: que el código fuera tan legible como el inglés.

> 💡 **Curiosidad:** El nombre no viene de la serpiente, sino del grupo de comedia británico *Monty Python*.

### **La Filosofía: "The Zen of Python"**

Python tiene su propia "constitución" interna. Si ejecutas `import this` en una consola de Python, verás los principios que guían su diseño. Los más importantes para nosotros son:

* *Lo bello es mejor que lo feo.*

* *Lo simple es mejor que lo complejo.*

* *La legibilidad cuenta.*

### **¿Cómo funciona por dentro? (Compilado vs Interpretado)**

Python es un lenguaje **interpretado de alto nivel**. A diferencia de C o C++ (que se compilan a código máquina directo para el procesador), Python funciona con una capa intermedia.

1. **Código Fuente (`.py`):** Lo que tú escribes.

2. **Compilación a Bytecode (`.pyc`):** El intérprete traduce tu código a un formato intermedio de bajo nivel (bytecode) para que sea más rápido de leer.

3. **Python Virtual Machine (PVM):** Es el motor real. Lee el bytecode y lo traduce a instrucciones máquina para tu sistema operativo específico (Windows, Linux, macOS).

#### ¿Por qué esto es genial para Hacking?

* **Portabilidad:** Escribes tu script de ataque en tu Kali Linux y funcionará en el servidor Windows de la víctima sin recompilar nada.

* **Rapidez de desarrollo:** No tienes que esperar a compilar. Escribes, ejecutas, fallas, corriges y repites al instante.

# **⭐ Características y ventajas**

### **Características principales**

* **Sintaxis limpia:** Usa **identación** (espacios) para definir bloques de código en lugar de llaves `{}`. Esto obliga a escribir código ordenado.

* **Tipado Dinámico (Duck Typing):** No necesitas declarar si una variable es `int` o `string`. Python lo deduce sobre la marcha.

  * *Lema:* "Si camina como un pato y hace cuack como un pato, entonces es un pato".


* **Multiparadigma:** No te obliga a una sola forma de trabajar.

  * *Imperativo:* Instrucciones paso a paso (scripts simples).

  * *Orientado a Objetos (POO):* Para herramientas complejas y modulares.

  * *Funcional:* Para procesamiento de datos avanzado.


* **"Batteries Included":** La biblioteca estándar es gigantesca. Tienes módulos nativos para casi todo (`socket` para redes, `os` para sistema, `re` para regex) sin instalar nada extra.

### **Ventajas en el mundo ofensivo ⚔️**

1. **Productividad brutal:** Un exploit que en C te llevaría 500 líneas y 3 horas, en Python lo haces en 50 líneas y 20 minutos.

2. **Librerías de Hacking:** La comunidad ha creado herramientas increíbles (`Scapy`, `Impacket`, `Requests`, `Pwntools`) que facilitan la vida.

3. **Scripting Glue:** Python es perfecto para "pegar" otros programas. Puedes hacer un script que ejecute Nmap, procese el resultado y lance un ataque con Metasploit automáticamente.

# **🔄 Diferencias Python 2 vs Python 3 y PIP**

Aunque Python 2 murió oficialmente en 2020, **en ciberseguridad sigue siendo relevante**. Te encontrarás muchos servidores antiguos o exploits viejos (en Exploit-DB) escritos en Python 2 que tendrás que arreglar.

### **Diferencias Clave (Cheat Sheet)**

| Característica | Python 2 (Legacy) | Python 3 (Actual) |
| --- | --- | --- |
| **Print** | `print "Hola"` (Sentencia) | `print("Hola")` (Función) |
| **Input** | `raw_input()` para texto | `input()` para texto |
| **División** | `5 / 2 = 2` (Entera) | `5 / 2 = 2.5` (Flotante) |
| **Strings** | ASCII por defecto | Unicode (UTF-8) por defecto |
| **Manejo de bytes** | Confuso (str y bytes mezclados) | Estricta separación `str` vs `bytes` (Vital en redes) |

> ⚠️ **Nota para Hackers:** Si encuentras un exploit antiguo en Internet y te da errores de sintaxis (`SyntaxError`), lo primero que debes revisar son los `print` (falta de paréntesis) y cómo maneja los bytes/strings.

### **El gestor de paquetes: PIP**

**PIP** (*Pip Installs Packages*) es la tienda de aplicaciones de Python. Se conecta al **PyPI** (Python Package Index) para descargar librerías de terceros.

* **`pip install <paquete>`**: Instala una librería.

* **`pip freeze`**: Muestra qué tienes instalado.

* **Entornos Virtuales (`venv`):** Es una mala práctica instalar todo en el Python global de tu sistema (puedes romper tu Linux). Lo ideal es crear entornos aislados para cada proyecto:

  ```bash
  python3 -m venv entorno_hacking
  source entorno_hacking/bin/activate
  ```
